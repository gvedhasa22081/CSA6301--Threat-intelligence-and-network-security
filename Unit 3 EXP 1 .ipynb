```python
import pandas as pd

feed1 = [
    {"ip":"192.168.1.10","type":"IP"},
    {"domain":"evil.com","type":"Domain"}
]

feed2 = [
    {"ip":"192.168.1.10","type":"IP"},
    {"domain":"bad.com","type":"Domain"}
]

ioc=[]

for item in feed1:
    if item not in ioc:
        ioc.append(item)

for item in feed2:
    if item not in ioc:
        ioc.append(item)

print(pd.DataFrame(ioc))
```
