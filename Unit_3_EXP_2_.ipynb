```python
import pandas as pd

logs = [
    ("Server1","10.0.0.1","Failed"),
    ("Server2","10.0.0.1","Failed"),
    ("Server3","10.0.0.1","Failed"),
    ("Server1","10.0.0.2","Success")
]

df=pd.DataFrame(logs,columns=["Server","IP","Status"])

failed=df[df["Status"]=="Failed"]

result=failed.groupby("IP").size()

print(result)

for ip,count in result.items():
    if count>=3:
        print("Brute Force Detected:",ip)
```
