```python
logs=[
{"user":"john","event":"login"},
{"user":"john","event":"failed"},
{"user":"john","event":"failed"},
{"user":"john","event":"failed"}
]

count=0

for log in logs:
    if log["event"]=="failed":
        count+=1

if count>=3:
    print("ALERT : Multiple Failed Logins")
```
